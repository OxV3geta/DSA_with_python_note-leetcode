#### Example 1: Temperature Conversion

```python
def temp_conv(temp,unit):
    if unit == 'C':
        return temp * 9/5+32 ## cecius to fahrenheit
    elif unit == 'F':
        return (temp-32)*5/9 ## faremheit to celsius
    else:
        return None
    
print(temp_conv(25,'C'))
print(temp_conv(77,'F'))
```

```python
def pass_checker(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    return True

print(pass_checker("WeakPwd"))
print(pass_checker("Str0ngPwd!"))
```

```python
def total_price(cart):
    total_cost = 0
    for items in cart:
        total_cost += items['price']*items['quantity']
    
    return total_cost
    
cart=[
    {'name':'Apple','price':5,'quantity':4},
    {'name':'Banana','price':2,'quantity':6},
    {'name':'Orange','price':8,'quantity':3}

]
sum = total_price(cart)
print(sum)
```

```python
def check_palindrom(s):
    s = s.lower().replace(" ",'')
    return s

print(check_palindrom('A man a plan a canal Panama'))
print(check_palindrom("Hello"))
```

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n*factorial(n-1)

print(factorial(5))
```

```python
def factorial(n):
    res = 1
    for i in range(1,n+1):
        res *= i
    return res

print(factorial(5))
```

```python
def count_word_frequency(file_path):
    word_count = {}
    with open(file_path,'r') as file:
        for line in file:
            words = line.split()
            for word in words:
                word=word.lower().strip('.,!?;:"\'')
                word_count[word] = word_count.get(word,0)+1
                
    return word_count

filepath='sample.txt'
word_frequency=count_word_frequency(filepath)
print(word_frequency)               
                
                
```
