```python
def even_or_odd(num):
    if num%2 == 0:
        print("Even")
    else:
        print("odd")   
    
```

```python
even_or_odd(23)
```

```python
###We use this type of argument(Positional arguments) for one main reason: Flexibility when you don’t know how many inputs a function will receive.

def print_numbers(*goku):
    for number in goku:
        print(number)
```

```python
print(6,9,69,6324,23,"bruhhhhhhhhhh")
```

```python
###In simple terms, Keyword Arguments allow you to pass inputs as name-value pairs (like a dictionary), so you always know what each value represents.

def print_details(**detail):
    for key,value in detail.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Krish",age="32",country="India")
```

```python
def details(*args,**kwargs):
    for val in args:
        print(f"positional argument:{val}")
    for key,value in kwargs.items():
        print(f"{key}:{value}")
        
```

```python
details(1,2,3,7,'bruh',name='vegeta',age='47',)
```

```python
## return
def multiply(a,b):
    return a*b

multiply(4,5)
```

```python
## return multiple parameters
def multiply(a,b):
    return (a*b,b)

multiply(6,7)
```
